### Grafik zu unübersichtlich. Weshalb mithilfe eines rollenden Fensters (größe 10 Zeilen), eine Aggregation stattffindet, mit Intention klarere Trends zu erkennen

```python
df["rolling_10"]=df["Deutschland"].rolling(window=10).mean()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Jahr"], df["Deutschland"], label="Original")
plt.title("Entwicklung der Temperatur im Januar")


plt.plot(df["Jahr"], df["rolling_10"], label="10 Year moving average")
plt.title("Entwicklung Temperature im Januar")
plt.xlabel("Jahr")
plt.ylabel("Temperatur")
plt.legend()
plt.show()
```

#### Erkenntnis: Die Temperatur scheint ab 1945 durchschnittlich höhere Werte zu haben

```python
latter=df[df["Jahr"]>1945]
early=df[df["Jahr"]<=1945]
```

```python
print(latter["Deutschland"].mean())
print(early["Deutschland"].mean())
```

#### Die durchschnittliche Januartemperatur in Deutschland liegt im Zeitraum ab 1945 etwa 1 °C höher als im Zeitraum vor 1945.

```python
bundeslaender=["Bayern", "Thueringen", "Sachsen", "Baden-Wuerttemberg", 
"Thueringen/Sachsen-Anhalt", "Brandenburg", "Brandenburg/Berlin", "Hessen", 
"Mecklenburg-Vorpommern", "Sachsen-Anhalt", 
"Rheinland-Pfalz", "Saarland", "Schleswig-Holstein", 
"Niedersachsen", "Niedersachsen/Hamburg/Bremen", "Nordrhein-Westfalen", "Deutschland"]
df[bundeslaender].mean().sort_values()
```

```python
plt.figure(figsize=(10,6))
df[bundeslaender].mean().sort_values().plot(kind="barh")
plt.show()
```

```python
fig, axes=plt.subplots(nrows=1, ncols=3,figsize=(10,5))

axes[0].boxplot(df["Bayern"])
axes[0].set_title("Verteilung Temperatur Bayern")
axes[0].set_label("Temperatur")
axes[0].grid(True, linestyle='--', alpha=0.5)

axes[1].boxplot(df["Nordrhein-Westfalen"])
axes[1].set_title("Verteilung Temperatur NRW")
axes[1].set_label("Temperatur")
axes[1].grid(True, linestyle='--', alpha=0.5)

axes[2].boxplot(df["Deutschland"])
axes[2].set_title("Verteilung Temperatur DE")
axes[2].set_label("Temperatur")
axes[2].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
